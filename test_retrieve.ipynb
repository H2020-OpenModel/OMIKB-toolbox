```python
g = rdflib.Graph()
```

```python
g.parse("http://www.w3.org/People/Berners-Lee/card")
```

```python
print(len(g))
```

```python
kb=kb_toolbox()
```

```python
# Loop over the triples and push each one to your Fuseki instance
for s, p, o in g:
    insert_query = f"""
    PREFIX rdf: <http://www.w3.org/1999/02/22-rdf-syntax-ns#>
    PREFIX rdfs: <http://www.w3.org/2000/01/rdf-schema#>

    INSERT DATA {{
        <{s}> <{p}> {f'<{o}>' if not isinstance(o, rdflib.Literal) else f'"{o}"'} .
    }}
    """
    
    response = kb.update(insert_query)
    
    
```
